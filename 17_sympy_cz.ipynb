# SymPy

- SymPy je knihovna jazyka Python pro symbolickou matematiku.

- Symbolické výpočty se zabývají výpočtem matematických objektů v symbolické podobě.
To znamená, že matematické objekty jsou reprezentovány přesně, nikoli přibližně, a matematické výrazy s nevyhodnocenými proměnnými zůstávají v symbolickém tvaru.

- Je pravda, že software Mathematica může být výkonnější.
SymPy je však skvělá open-source alternativa, která pro mnoho úloh postačuje.
SymPy lze navíc snadno propojit s dalšími knihovnami Pythonu, což z něj činí všestranný nástroj pro vědecké výpočty.
Můžete například připravit složitý symbolický výraz v SymPy a poté použít NumPy k rychlému numerickému vyhodnocení nebo Matplotlib k vizualizaci výsledků.

# Symbolické proměnné a výrazy v SymPy

- Při symbolickém výpočtu zůstávají matematické výrazy v přesném symbolickém tvaru, zatímco numerický výpočet je aproximuje pomocí čísel s plovoucí desetinnou čárkou.

```python
import sympy
import math
print(math.sqrt(8))
print(sympy.sqrt(8))
```

- Výrazy v SymPy se vytvářejí pomocí symbolických proměnných, které lze definovat funkcí `symbols`. V jejím řetězcovém argumentu jsou proměnné odděleny mezerami.

- Poznámka: Symbolické proměnné se ve výchozím nastavení považují za komplexní čísla.

```python
import sympy
x, y = sympy.symbols('x y')
expr = x + 2*y
print(expr)
```

- Mezi symbolickou proměnnou SymPy a běžnou proměnnou Pythonu, která na ni odkazuje, je rozdíl. Následující příklad jej ilustruje:

```python
import sympy
x, y = sympy.symbols('y x')
expr = x + 2*y
print(expr)
```

- Předchozí kód ukazuje, že symbolické proměnné SymPy mohou mít jiná jména než proměnné Pythonu, které na ně odkazují. Běžně se však kvůli přehlednosti používají pro obě stejná jména.

- Symbolické proměnné mohou mít také delší názvy:

```python
import sympy
mu, nu, theta = sympy.symbols('mu nu theta')
expr = (mu + nu) ** (2 - theta)
print(expr)
```

- Výrazy jsou neměnné, což znamená, že každá operace s výrazem vytvoří nový výraz, aniž by změnila původní. Například sečtením dvou výrazů vznikne nový výraz a původní zůstanou beze změny.

```python
import sympy
x, y = sympy.symbols('x y')
expr1 = x + y
expr2 = x - y
expr3 = expr1 + expr2
print('expr1: ', expr1)
print('expr2: ', expr2)
print('expr3: ', expr3)
```

- Řetězce lze převést na výrazy SymPy pomocí funkce `sympify`. To se hodí při načítání výrazů z externích zdrojů, například ze souborů nebo od uživatele.

```python
import sympy
expr = sympy.sympify('x**2 + 2*x + 1')
print(expr)
```

- Hezkého formátování výrazů dosáhnete zavoláním funkce `sympy.init_printing()`.

```python
import sympy
sympy.init_printing()
x, y = sympy.symbols('x y')
expr = x + y ** 2
expr # Do not use print() here to see pretty output
```

# Numerické vyhodnocování výrazů SymPy

- Výrazy SymPy lze numericky vyhodnotit pomocí metody `evalf`.

- Metoda `evalf` vypočítá číselnou aproximaci výrazu s určeným počtem desetinných míst, který udává parametr `n`.

- Chcete-li výraz vyhodnotit pro konkrétní hodnoty proměnných, použijte parametr `subs`.

```python
import sympy
x, y = sympy.symbols('x y')
expr = x + y
num_expr = expr.evalf(subs = {x : 1.5, y : 2.5}, n = 5)
print(num_expr)
```

```python
# Calculate pi to 100 decimal places
import sympy
print(sympy.pi.evalf(n = 100))
```

# Funkce lambdify pro rychlé numerické vyhodnocování

- Funkce `evalf` se hodí pro jednoduché vyhodnocení v jednom bodě, ale pro vyhodnocování výrazu v mnoha bodech existuje účinnější způsob. Například při vyhodnocení v tisíci bodech by byl SymPy zbytečně pomalý, zvláště pokud vám stačí přesnost stroje. Místo něj použijte knihovny jako NumPy.

- Nejjednodušší způsob převodu výrazu SymPy na výraz vhodný k numerickému vyhodnocování je funkce `lambdify()`. `lambdify()` funguje podobně jako funkce lambda, ale názvy ze SymPy převádí na názvy používané zadanou numerickou knihovnou, například NumPy.


```python
import sympy
import numpy
x = sympy.symbols('x')
expr = x ** 2 + 1
f = sympy.lambdify(x, expr, "numpy")
f(numpy.array([0, 1, 2, 3, 4, 5]))
```

# Rozvoj výrazů

- Funkce `expand()` roznásobuje součiny a mocniny.

- Pro konkrétní druhy rozvoje existují také specializované funkce, například `expand_trig()` pro goniometrické výrazy a `expand_log()` pro logaritmické výrazy.

- Další podrobnosti najdete na stránce [https://docs.sympy.org/latest/tutorials/intro-tutorial/simplification.html](https://docs.sympy.org/latest/tutorials/intro-tutorial/simplification.html).

```python
import sympy
x = sympy.symbols('x')
expr = (x + 1) ** 2
expanded_expr = sympy.expand(expr)
expanded_expr
```

```python
# expand_trig() example
import sympy
x = sympy.symbols('x')
expr = sympy.sin(x + sympy.pi / 4)
expanded_expr = sympy.expand_trig(expr)
expanded_expr
```

- Pokud rovnost neplatí pro všechna komplexní čísla, k rozvoji nedojde. Například rovnost $\log(x y) = \log(x) + \log(y)$ neplatí kvůli nespojitosti logaritmu v hodnotě $2\pi$ (např. $\log(-i) = -i\pi/2 \neq \log(-1) + \log(i)$). Pro kladné reálné hodnoty $x$ a $y$ však platí vždy. K rozvoji tedy dojde pouze tehdy, pokud určíme, že symbolické proměnné jsou kladné.

```python
import sympy
x, y = sympy.symbols('x y')                    # the logarithm expansion will not happen
# x, y = sympy.symbols('x y', positive = True) # the logarithm expansion will happen
expr = sympy.log(x * y)
expanded_expr = sympy.expand_log(expr)
expanded_expr
```

# Zjednodušování výrazů

- Funkce `simplify()` se pokouší zjednodušit výrazy do kompaktnější podoby.

- Pro konkrétní druhy zjednodušení existují specializované funkce, například `trigsimp()` pro goniometrické výrazy, `logcombine()` pro logaritmické výrazy a `factor()` pro rozklad polynomů.

- Poznámka: `simplify()` může být zbytečně pomalá, protože před výběrem nejlepšího výsledku zkouší mnoho druhů zjednodušení. Pokud přesně víte, jaké zjednodušení potřebujete, je lepší použít příslušnou specializovanou funkci či funkce.

- Další podrobnosti najdete na stránce [https://docs.sympy.org/latest/tutorials/intro-tutorial/simplification.html](https://docs.sympy.org/latest/tutorials/intro-tutorial/simplification.html).

```python
import sympy
x = sympy.symbols('x')
expr = sympy.sin(x) ** 2 + sympy.cos(x) ** 2
simpl_expr = sympy.simplify(expr)
simpl_expr
```

```python
# trigsimp() example
import sympy
x = sympy.symbols('x')
expr = sympy.sin(x) ** 2 - sympy.cos(x) ** 2
simpl_expr = sympy.trigsimp(expr)
simpl_expr
```

```python
# logcombine() example
import sympy
x = sympy.symbols('x', positive = True) # specify that x is positive to avoid issues with logarithm domain; otherwise, simplification may not proceed
expr = sympy.log(x) + sympy.log(x + 1)
simpl_expr = sympy.logcombine(expr)
simpl_expr
```

```python
# factor() example
import sympy
x = sympy.symbols('x')
expr = x**2 - 1
factored_expr = sympy.factor(expr)
factored_expr
```

# Derivace

- Funkce `diff()` počítá derivace výrazů podle zadaných proměnných.

- Funkci `diff()` a metodu `diff()` výrazů lze používat zaměnitelně.

```python
# diff() function example
import sympy
x, y = sympy.symbols('x y')
expr = x**2 * y + sympy.sin(x * y)
deriv_expr = sympy.diff(expr, x)
deriv_expr
```

- Stejnou funkcionalitu nabízí také metoda `diff()` výrazů.

```python
# diff() method example
import sympy
x, y = sympy.symbols('x y')
expr = x**2 * y + sympy.sin(x * y)
deriv_expr = expr.diff(x)
deriv_expr
```

- Lze vypočítat i několik derivací najednou. V následujícím příkladu počítáme třetí derivaci podle `x` a druhou derivaci podle `y`.

```python
import sympy
x, y = sympy.symbols('x y')
expr = x**2 * y + sympy.sin(x * y)
deriv_expr = expr.diff(x, x, x, y, y)
deriv_expr
```

- Derivaci neurčeného řádu lze vytvořit pomocí n-tice `(x, n)`, kde `n` udává řád derivace podle `x`.

```python
import sympy
x, n = sympy.symbols('x n')
expr = x**2 * sympy.sin(x)
deriv_expr = expr.diff((x, n))
deriv_expr
```

- Pomocí třídy `Derivative` lze s operátorem derivování pracovat jako se symbolickým objektem. Derivace tak můžeme upravovat, aniž bychom je hned vyhodnotili.

```python
import sympy
x, y = sympy.symbols('x y')
expr = x**2 * y + sympy.sin(x * y)
deriv_expr = sympy.Derivative(expr, x, x, x)
deriv_expr
```

- Odloženou derivaci lze vyhodnotit metodou `doit()`.

```python
deriv_expr = deriv_expr.doit()
deriv_expr
```

# Integrály

- K výpočtu integrálu použijte funkci `integrate()`.

- Existují dva druhy integrálů: určité a neurčité.

- Neurčitý integrál: ve funkci `integrate()` uveďte za výrazem proměnnou.

```python
import sympy
x = sympy.symbols('x')
expr = x
int_expr = sympy.integrate(expr, x)
int_expr
```

- Všimněte si, že SymPy vynechává integrační konstantu.

- Stejnou funkcionalitu nabízí také metoda `integrate()` výrazů.

```python
import sympy
x = sympy.symbols('x')
expr = x
int_expr = expr.integrate(x)
int_expr
```

- Určitý integrál: funkci `integrate()` předejte proměnnou a meze jako n-tici.

```python
import sympy
x = sympy.symbols('x')
expr = x
int_expr = expr.integrate((x, 0, 1))
int_expr
```

- Symbol pro nekonečno je `sympy.oo`. Zápis oo připomíná $\infty$ a snadno se píše.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.exp(-x)
int_expr = expr.integrate((x, 0, sympy.oo))
int_expr
```

- Stejně jako u `Derivative` můžete pomocí `Integral` vytvořit nevyhodnocený integrál. Později jej vyhodnotíte zavoláním `doit()`.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.exp(-x)
int_expr = sympy.Integral(expr, x)
int_expr
```

```python
int_expr.doit()
```

# Numerická integrace

- SymPy podporuje také numerickou integraci. Využívá přitom možnosti knihovny mpmath pracovat s vysokou přesností, čímž zvyšuje přesnost numerických výpočtů integrálů.

- Numerickou integraci provedete metodou `evalf()` nad objektem `Integral`. Počet platných číslic lze nastavit parametrem `n`.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.exp(-x)
int_expr = sympy.Integral(expr, (x, 0, 1))
int_expr.evalf(30)
```

- Integrand lze také převést pomocí lambdify a pro potenciálně rychlejší výpočet složitých integrálů použít metody numerické integrace z knihoven, jako je SciPy.

```python
import sympy
import scipy.integrate
x = sympy.symbols('x')
expr = sympy.exp(-x)
f = sympy.lambdify(x, expr, 'scipy')
print(scipy.integrate.quad(f, 0, 1))
```

- Všimněte si, že `scipy.integrate.quad` vrací n-tici obsahující hodnotu integrálu a odhad absolutní chyby.

# Limity

- SymPy umí počítat symbolické limity funkcí `limit()`. Syntaxe pro výpočet $$\lim_{x \to a} f(x)$$ je `sympy.limit(expr, x, a)`.

```python
import sympy
x = sympy.symbols('x')
expr = sympy.sin(x) / x
limit_expr = sympy.limit(expr, x, 0)
limit_expr
```

- Chcete-li vypočítat jednostrannou limitu, předejte funkci `limit()` jako čtvrtý argument `'+'` nebo '-'. Například pro výpočet $$\lim_{x \to 0^+} 1/x$$ použijte `sympy.limit(1/x, x, 0, '+')`.

```python
import sympy
x = sympy.symbols('x')
expr = 1 / x
limit_expr = sympy.limit(expr, x, 0, '+')
limit_expr
```

- Obdobně postupujte pro limitu z druhé strany:

```python
import sympy
x = sympy.symbols('x')
expr = 1 / x
limit_expr = sympy.limit(expr, x, 0, '-')
limit_expr
```

# Rovnice

- SymPy řeší rovnice pomocí funkcí `solveset()`, `linsolve()`, `nonlinsolve()`, `root()`, `dsolve()` a dalších.

- Další podrobnosti a omezení najdete na stránce [https://docs.sympy.org/latest/tutorials/intro-tutorial/solvers.html](https://docs.sympy.org/latest/tutorials/intro-tutorial/solvers.html).

- Objekty rovnic lze vytvořit funkcí `Eq()`. Poté se předají řešící funkci spolu s proměnnou či proměnnými, pro které rovnici řešíme.

```python
import sympy
x = sympy.symbols('x')
eq = sympy.Eq(x**2, 1) # Syntax: Eq(left_side, right_side)
sympy.solveset(eq, x)
```

- Řešící funkci lze předat přímo také výraz; v takovém případě se předpokládá, že se rovná nule. Následující příklad je ekvivalentní předchozímu:

```python
import sympy
x = sympy.symbols('x')
expr = x**2 - 1  # Expression assumed to be equal to zero
sympy.solveset(expr, x)
```

# Soustavy rovnic

- SymPy umí řešit soustavy lineárních rovnic pomocí funkce `linsolve()`.

```python
import sympy
x, y, z = sympy.symbols('x y z')
system = [x + y + z - 1,
          x + y + 2*z - 3 ]
sympy.linsolve(system, (x, y, z))
```

- Nelineární soustavy rovnic se řeší pomocí `nonlinsolve()`.

```python
import sympy
x, y = sympy.symbols('x y')
system = [x**2 - 2*y**2 -2,
          x*y - 2]
sympy.nonlinsolve(system, (x, y))
```

# Diferenciální rovnice

- SymPy umí řešit obyčejné diferenciální rovnice (ODE) pomocí funkce `dsolve()`.

- Nejprve vytvořte neurčenou funkci tak, že funkci `symbols()` předáte argument `cls = sympy.Function`.

```python
import sympy
f, g = sympy.symbols('f g', cls = sympy.Function)
```

- Poté sestavte diferenciální rovnici pomocí metody `diff()` neurčené funkce a předejte ji funkci `dsolve()` spolu s funkcí, kterou chcete určit.

```python
diffeq = sympy.Eq(f(x).diff(x, x) - 2 * f(x).diff(x) + f(x), sympy.sin(x))
sympy.dsolve(diffeq, f(x))
```
