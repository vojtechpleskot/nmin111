Cíl: Seřadit čísla (nebo jiné prvky) podle předem daného pořadí (např. vzestupně).

Cíl je jednoduchý, ale řešení už ne, protože časová složitost může být vysoká.
Nejjednodušší řadicí algoritmy (např. bubble sort) mají časovou složitost O(n^2), zatímco nejpokročilejší algoritmy dosahují O(n log n).

# Nejjednodušší řadicí algoritmy

Podívejme se na jednoduché řadicí algoritmy s časovou složitostí O(n^2).

### Řazení opakovaným výběrem minima


```python
# Sorting by repeated selection of minimum

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        # Assume the minimum is the first element of the unsorted part
        min_index = i
        # Iterate through the unsorted elements
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        # Swap the found minimum element with the first element of the unsorted part
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr

# Example usage
numbers = [64, 25, 12, 22, 11, -5]
sorted_numbers = selection_sort(numbers)
print("Sorted array:", sorted_numbers)
```

```python
# Let's experiment with high-n inputs
N = 10000
long_list = [i for i in range(N, 0, -1)]

# Set the timer.
import time
start_time = time.time()

# Sort the list using our selection sort implementation
sorted_long_list = selection_sort(long_list)

# Stop the timer.
end_time = time.time()

# Display the results.
print("Sorted large array:", sorted_long_list)
print("Time taken:", end_time - start_time)
```

# Řazení řetězců: lexikografické pořadí

- Řetězce se řadí podle kódových bodů Unicode svých znaků.

- Kódové body Unicode „běžných“ znaků jsou přiřazeny intuitivně (např. 'a' < 'b' < 'c', 'A' < 'B' < 'C', '0' < '1' < '2').

- Velká písmena mají nižší kódové body Unicode než malá písmena (např. 'A' < 'a').

- Nealfanumerické znaky mají různé kódové body Unicode, což může ovlivnit pořadí při řazení.

- Pokud jsou první znaky dvou řetězců stejné, porovnají se druhé znaky a tak dále.

```python
# Sort list of names in the lexicographical order.
names = ['Charlie', 'Alice', 'Bob', 'David', 'Bill']
sorted_names = sorted(names)
print(sorted_names)
```

# Řazení podle vlastního klíče

- Pomocí parametru `key` funkce `sorted()` nebo metody `.sort()` určete vlastní kritérium řazení.

- Například můžete seřadit seznam řetězců podle jejich délky.

- Parametr `key` přijímá funkci, která z každého prvku kolekce získá klíč pro porovnání.


```python
names = ['Charlie', 'Alice', 'Bob', 'David', 'Bill']
sorted_names = sorted(names, key = len)
print(sorted_names)  # Output: ['Bob', 'Bill', 'Alice', 'David', 'Charlie']
```

```python
# One can sort according to several criteria by using a tuple as the key.
# The lexicographical order is followed in the sense that the first tuple elements are compared first,
# and only if they are equal, the second elements are compared, etc.
people = [['Alice', 30], ['Bob', 25], ['Charlie', 30], ['David', 20]]

# Define the key function.
def sort_key(person):
    name = person[0]
    age = person[1]
    return (age, name)

# Sort by age, then by name
sorted_people = sorted(people, key = sort_key)
print(sorted_people) # Output: [('David', 20), ('Bob', 25), ('Alice', 30), ('Charlie', 30)]
```

```python
# Sort the people above from the oldest to the youngest; sort people with the same age according to their name alphabetically.
def sort_key_desc(person):
    name = person[0]
    age = person[1]
    return (-age, name)

sorted_people_desc = sorted(people, key = sort_key_desc)
print(sorted_people_desc)  # Output: [('Alice', 30), ('Charlie', 30), ('Bob', 25), ('David', 20)]
```

### Funkce lambda jako řadicí klíče

- Funkce lambda můžete použít jako parametr `key` a definovat tak vlastní kritéria řazení přímo na místě.

```python
# Define the sort_key function using a lambda function.
sorted_people = sorted(people, key = lambda person: (person[1], person[0]))
print(sorted_people) # Output: [('David', 20), ('Bob', 25), ('Alice', 30), ('Charlie', 30)]
```
