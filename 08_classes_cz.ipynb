# Třídy

- Python podporuje objektově orientované programování (OOP).
- Třídy představují typy z „reálného světa“ (např. pes, auto nebo člověk).
- Objekty jsou konkrétními instancemi tříd.
    - Například pes Fido, kterého má moje babička, je objektem třídy Dog.
- Třídy mají atributy, což jsou proměnné přiřazené ke třídě. Tyto proměnné uchovávají data o objektu.
    - Každého psa určují například atributy jako jméno, barva, výška nebo hmotnost.
- Třídy mají také metody, tedy funkce přiřazené ke třídě, které pracují s daty uloženými v atributech.
    - Pes může mít metodu eat(), která zvýší jeho atribut hmotnosti.

# Definice třídy

- Třída se definuje pomocí klíčového slova `class`, za kterým následuje název třídy a dvojtečka.
- Tělo třídy obsahuje její atributy a metody.
- Typická třída má metodu `__init__`, která se zavolá při vytvoření objektu této třídy.
- Atributy objektu se inicializují v metodě `__init__`.

```python
# Class definition example
class Dog:

    def __init__(self, name, color, height, weight):
        self.name   = name
        self.color  = color
        self.height = height
        self.weight = weight
        self.x      = 0
        self.y      = 0

    def eat(self, food_weight):
        self.weight += food_weight

    def move(self, dx, dy):
        self.x += dx
        self.y += dy

    def bark(self):
        print("Woof!")

# Creating an instance of the Dog class
fido = Dog("Fido", "brown", 50, 20)

# Feeding and walking Fido
fido.eat(5)
fido.move(10, 15)

# When Fido meets another dog
fido.bark()
```

```python
# Checking Fido's position
print(fido.x, fido.y)
```

```python
# It's possible to change attributes directly.
fido.x += 1
fido.y += 3
print(fido.x, fido.y)
```

Poznámky:

- Veškerý kód třídy musí být odsazený.
- Podle konvence se názvy tříd zapisují ve formátu CamelCase.
- Prvním parametrem každé metody třídy je `self`, který odkazuje na samotný objekt.
    - Při volání metody se samotný objekt automaticky předá jako parametr `self`.
- Při vytvoření objektu se metoda `__init__` zavolá automaticky.
- Metody se volají pomocí tečkové notace: `object.method()`.
- K atributům se také přistupuje pomocí tečkové notace: `object.attribute`.
- Atributy lze upravovat pomocí tečkové notace: `object.attribute = value`.



```python
# Function that "steals" a dog by changing its position.
# Note how Fido's attributes are modified directly.
def steal_a_dog(dog, new_x, new_y):
    dog.bark()
    dog.x = new_x
    dog.y = new_y

# Stealing Fido and printing his new position.
steal_a_dog(fido, 1000, 1000)
print(fido.x, fido.y)
```

Poznámky k funkci `steal_a_dog`:

- Je definována mimo třídu, protože nepopisuje chování psů.
- Funkce se velmi podobá metodám třídy Dog. Parametr `dog` zde zastává roli parametru `self` v metodách třídy. Při volání funkce mu však musíme hodnotu předat výslovně.

# Dědičnost

V reálném světě existují hierarchie typů.
Například různá plemena psů sdílejí společné atributy a metody, ale mají také některé specifické.
Dědičnost umožňuje takové hierarchie modelovat.

Nová třída, například `Dachshund`, může dědit atributy a metody z existující třídy.
Nová třída se nazývá podtřída a existující třída nadtřída.
Podtřída může přidávat nové atributy a metody nebo přepisovat ty existující.

Třída `Dachshund` má všechny atributy a metody třídy `Dog`, potřebuje však i některé specifické atributy.
Délka je pro jezevčíka rozhodně důležitou vlastností.
Třída `Dachshund` proto musí mít další atribut `length`.
Jezevčíci také mohou štěkat specifickým způsobem, který se liší od ostatních psů.

```python
# Define the Dachshund subclass
class Dachshund(Dog):

    def __init__(self, name, color, height, weight, length):
        super().__init__(name, color, height, weight)
        self.length = length

    def bark(self):
        print("Woof! Woof! (in a high-pitched voice)")

# Creating an instance of the Dachshund subclass
# Note the additional argument,
# corresponding to the length parameter in the new __init__ method.
noodle = Dachshund("Noodle", "brown", 25, 10, 50)

# Noodle barks in a different way than Fido.
noodle.bark()
```

Poznámky:

- Metodu `__init__` je třeba v podtřídě znovu definovat, aby inicializovala nový atribut.
    - Funkce `super()` slouží k zavolání metody `__init__` nadtřídy, která inicializuje zděděné atributy.
- Metoda `bark` je v podtřídě přepsána, aby poskytovala specifickou implementaci pro jezevčíky.

# Dokumentace tříd

Ke třídám a metodám lze přidat dokumentační řetězce (docstringy), které popisují jejich účel a použití.
Uzavírají se do trojitých uvozovek a umisťují se bezprostředně za definici třídy nebo metody.
Dokumentační řetězce tříd mohou obsahovat následující části:
- Stručné shrnutí: jednořádkový popis třídy
- Rozšířené shrnutí: podrobnější popis třídy
- Parametry: argumenty konstruktoru `__init__`
- Atributy: seznam veřejných atributů instance se stručnými popisy
- Metody: seznam veřejných metod se stručnými popisy
- Příklady: ukázky kódu demonstrující použití třídy (nejběžnější způsoby vytváření a používání jejích objektů)
- Další části podle potřeby (atributy třídy, poznámky, viz také...)

```python
# Example of docstrings in class and its methods

class Dog:
    """
    A class representing a dog.

    This class provides basic attributes and methods to model a dog.
    It can serve as a base class for more specific dog breeds.

    Parameters
    ----------
    name : str
        The name of the dog.
    color : str
        The color of the dog.
    height : float
        The height of the dog in centimeters.
    weight : float
        The weight of the dog in kilograms.

    Attributes
    ----------
    name : str
        The name of the dog.
    color : str
        The color of the dog.
    height : float
        The height of the dog in centimeters.
    weight : float
        The weight of the dog in kilograms.
    x : float
        The x-coordinate of the dog's position.
    y : float
        The y-coordinate of the dog's position.

    Methods
    -------
    eat(food_weight)
        Feeds the dog, increasing its weight by food_weight.
    move(dx, dy)
        Moves the dog by dx and dy in the x and y directions.
    bark()
        Makes the dog bark.

    Examples
    --------
    >>> my_dog = Dog("Buddy", "golden", 60, 30)
    >>> my_dog.eat(2)
    >>> my_dog.move(5, 10)
    >>> my_dog.bark()
    Woof!
    """

    def __init__(self, name, color, height, weight):
        self.name   = name
        self.color  = color
        self.height = height
        self.weight = weight
        self.x      = 0
        self.y      = 0

    def eat(self, food_weight):
        """
        Feeds the dog, increasing its weight by food_weight.

        Parameters:
        -----------
        food_weight : float
            The weight of the food in kilograms.
        """
        self.weight += food_weight

    def move(self, dx, dy):
        """
        Moves the dog by dx and dy in the x and y directions.

        Parameters:
        -----------
        dx : float
            The distance to move in the x direction.
        dy : float
            The distance to move in the y direction.
        """
        self.x += dx
        self.y += dy

    def bark(self):
        """
        Makes the dog bark.

        Prints "Woof!" to the console.
        """
        print("Woof!")
```

# Atributy instance a atributy třídy

Atributy instance jsou specifické pro každý objekt třídy.
Definují se v metodě `__init__` pomocí parametru `self`.
Každý objekt má vlastní kopii atributů instance.

Atributy třídy jsou naopak sdílené všemi instancemi dané třídy.
Definují se přímo v těle třídy, mimo všechny její instanční metody.

```python
# Example of class and instance attributes
class DogBase:
    sound = "Woof!"

    def bark(self):
        print(self.sound)

rex = DogBase()
fido = DogBase()
print(rex.sound, fido.sound) # Woof! Woof!
rex.sound = "Grrr!"
print(rex.sound, fido.sound) # Grrr! Woof!
DogBase.sound = "Bark!"
print(rex.sound, fido.sound) # Grrr! Bark!
```

# Protokoly

- Operátory jako ==, <, >, +, -, * mohou pracovat s objekty uživatelsky definovaných tříd.

- Toho se dosahuje definováním speciálních metod ve třídě.

- Například metoda `__eq__` se zavolá při použití operátoru ==.

    - Pokud je ve třídě definována metoda `__eq__`, výraz `obj1 == obj2` se převede na `obj1.__eq__(obj2)`.

```python
# Let's improve our Dog class
class Dog:

    def __init__(self, name, color, height, weight):
        self.name     = name
        self.color    = color
        self.height   = height
        self.weight   = weight
        self.x        = 0
        self.y        = 0
        self.consumed = []

    def eat(self, food, food_weight):
        self.consumed.append(food)
        self.weight += food_weight

    def move(self, dx, dy):
        self.x += dx
        self.y += dy

    def bark(self):
        print("Woof!")

    def __eq__(self, other):
        return (self.name == other.name and
                self.color == other.color and
                self.height == other.height and
                self.weight == other.weight)
    
    def __lt__(self, other):
        return self.weight < other.weight
    
    def __gt__(self, other):
        return self.weight > other.weight
    
    def __contains__(self, item):
        return item in self.consumed
    
    def __str__(self):
        return self.name
```

```python
# Example usage
fido = Dog("Fido", "brown" , 50, 20)
rex  = Dog("Rex" , "golden", 60, 30)

# __eq__ usage
print(fido == rex)  # Output: False

# __lt__ and __gt__ usage
print(fido < rex)   # Output: True
print(fido > rex)   # Output: False

# __contains__ usage
fido.eat("ham", 2)
print("ham" in fido)  # Output: True
print("cheese" in fido) # Output: False

# __str__ usage
print(fido)          # Output: Fido
```
