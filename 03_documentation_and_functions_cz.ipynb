# Funkce

- Opakovaně použitelné části kódu
- Vytvářejí definovaný výstup nebo provádějí konkrétní úlohu na základě zadaného vstupu
- Za názvem funkce následují závorky
    - Tomu se říká „volání funkce“
- Užitečné funkce:
    - print()
    - input()
    - abs(x), max()
    - print()
    - int(x), float(x)
    - math.sin(x)
- Definice vlastní funkce:
    ```python
    def function_name(parameter1, parameter2):
        ...
        return return_value
    ```
- Poznámka:
    - Klíčové slovo `def` musí být na začátku definice funkce
    - Název funkce volí uživatel
        - Bez mezer, nesmí začínat číslem
        - Měl by být výstižný a přiměřeně krátký
        - Dodržujte konvence pojmenování
            - Zpravidla malými písmeny a s podtržítky mezi slovy
    - Za názvem funkce následují závorky `()`
    - Parametry
        - Uvádějí se do závorek
        - Jsou nepovinné
        - Seznam názvů proměnných oddělených čárkami
        - Slouží jako zástupné symboly pro hodnoty předané funkci
    - Příkaz `return` ukončí vykonávání funkce a určí vracenou hodnotu
        - Pokud chybí, funkce ve výchozím nastavení vrátí `None`
    - Odsazení je důležité
        - Tělo funkce musí být odsazeno vůči řádku s `def`


````python
# Example function
def multiply_by_two(x):
    return x * 2

# Call the function and print the result.
result = multiply_by_two(5)
print(result)
````

````python
# More complex function

def solve_quadratic(a, b, c):
    """
    Returns the two solutions of a quadratic equation.

    The equation is of the form ax^2 + bx + c = 0,
    where a, b, and c are real numbers and a != 0.

    Parameters
    ----------
    a : float
        Coefficient of x^2
    b : float
        Coefficient of x
    c : float
        Constant term

    Returns
    -------
    tuple
        A tuple containing the two solutions if they exist.
    None
        None if there are no real solutions.

    Examples
    --------
    >>> solve_quadratic(1, -3, 2)
    (2.0, 1.0)
    """

    # Calculate the discriminant.
    discriminant = b ** 2 - 4 * a * c

    # Check if the discriminant is non-negative.
    if discriminant < 0:
        return None  # No real solutions
    
    # Calculate the two solutions.
    root1 = (-b + discriminant ** 0.5) / (2 * a)
    root2 = (-b - discriminant ** 0.5) / (2 * a)

    # Return the two solutions.
    return root1, root2
````

Komentáře:
- Řetězec bezprostředně za řádkem `def` se nazývá docstring
    - Obsahuje podrobný popis funkce
    - Je nepovinný, ale důrazně doporučený
    - Nástroje pro generování dokumentace jej mohou použít k vytvoření online referenčních příruček
    - Styl není přesně stanoven, existují však konvence
        - Viz například [PEP 257](https://peps.python.org/pep-0257/)
        - Výše uvedený příklad se řídí stylem docstringů NumPy
    - Docstring může být delší než samotný kód funkce
- Parametry `a`, `b` a `c` zastupují hodnoty předané funkci
    - Při volání funkce se hodnoty zadané jako argumenty přiřadí těmto parametrům
    - Parametry fungují uvnitř funkce jako lokální proměnné
        - Totéž platí pro pomocné proměnné definované uvnitř funkce (`discriminant`, `root1`, `root2`)
- Funkce může vracet několik hodnot
    - Oddělují se čárkami

# Výchozí hodnoty parametrů

Parametrům lze přiřadit výchozí hodnoty.
- Výchozí hodnoty se uvádějí v definici funkce.
- Parametry s výchozími hodnotami musí následovat až po parametrech bez výchozích hodnot.
- Pokud volající nezadá hodnotu parametru s výchozí hodnotou, použije se tato výchozí hodnota.

````python
# Example of a function whose parameter has a default value.
# The parameter `message` has a default value of "Hello".
# The parameter `message` must come after the parameter `name`, which has no default value.
def greet(name, message = "Hello"):
    print(f"{message}, {name}!")

# Calling the function with only the required parameter.
# The parameter `message` will take its default value.
greet("Alice")

# Calling the function with both parameters.
greet("Bob", "Hi")
````

# Poziční a pojmenované argumenty

Při volání funkce lze argumenty předat dvěma způsoby:

- Poziční argumenty: hodnoty se přiřazují parametrům podle jejich pozice ve volání funkce.
- Pojmenované argumenty: hodnoty se přiřazují parametrům tak, že se při volání funkce výslovně uvede název parametru.

Používáte-li při volání funkce poziční i pojmenované argumenty, poziční argumenty musí být uvedeny před pojmenovanými a musí dodržovat pořadí parametrů v definici funkce. Na pořadí pojmenovaných argumentů nezáleží.

````python
# Options for calling the function `greet`:

# 1. Positional arguments for `name` and `message`
greet("Alice", "Hi")

# 2. Keyword arguments for `name` and `message`
greet(name = "Charlie", message = "Hi")

# 3. Positional argument for `name` and keyword argument for `message`
greet("Bob", message = "Hello")
````

# Eukleidův algoritmus pro největší společný dělitel (GCD)

Algoritmus vychází z následujících skutečností:
- gcd(x, x) = x
- gcd(x, y) = gcd(y, x)
- gcd(x, y) = gcd(x - y, y) if x > y
