# Datové struktury

Mezi nejdůležitější datové struktury v Pythonu patří:
- Seznam (list)
- N-tice (tuple)
- Slovník (dictionary)
- Množina (set)

Slouží k ukládání dat – lze na ně nahlížet jako na datové kontejnery.
V těchto datových strukturách lze ukládat i velké množství dat.
Každá z těchto čtyř datových struktur má své vlastnosti a způsoby použití, díky nimž se hodí pro různé typy úloh. Pojďme si každou z nich podrobněji představit.


# N-tice

N-tice je kolekce podobná seznamu, ale je neměnná (po vytvoření ji nelze změnit).
Definuje se tak, že se posloupnost položek oddělených čárkami uzavře do kulatých závorek `()`.

```python
my_tuple = (1, 2, 3, 'apple', 'banana', 3.14, True)
```

Poznámka: Chcete-li vytvořit n-tici s jedinou položkou, přidejte na konec čárku:

```python
single_item_tuple = (42,)
```

````python
# To stress the immutable nature of tuples, let's generate an error
single_tuple = (42,)
# single_tuple[0] = 43  # This will raise a TypeError
````

Při přiřazování hodnot více proměnným se na pravé straně automaticky vytvoří n-tice (tuple packing):

```python
x, y, z = 1, 2, 3
```

Když funkce vrací více hodnot, ve skutečnosti vrací n-tici:

```python
def get_coordinates():
    return 10, 20
x, y = get_coordinates()
```

# Slovník

- Kolekce dvojic klíč-hodnota.
- Je neuspořádaný a měnitelný.
- Neumožňuje duplicitní klíče.
- Definuje se uzavřením posloupnosti dvojic klíč-hodnota oddělených čárkami do složených závorek `{}`.
    - Klíč je od hodnoty oddělen dvojtečkou `:`. Klíč je vlevo a hodnota vpravo.
    - Klíče musí být neměnného typu (např. řetězce, čísla nebo n-tice).

```python
phonebook = {
    'Alice'  : '123-456-7890',
    'Bob'    : '987-654-3210',
    'Charlie': '555-555-5555',
}
```

````python
# Create a dictionary representing a simple phonebook
phonebook = {
    'Alice'  : '123-456-7890',
    'Bob'    : '987-654-3210',
    'Charlie': '555-555-5555',
}
````

````python
# Values are accessed by their keys, using square brackets
bobs_number = phonebook['Bob']
print(bobs_number)  # Output: 987-654-3210
````

````python
# Values can be updated by assigning a new value to an existing key
phonebook['Alice'] = '111-111-1111'
print(phonebook)
````

````python
# New key-value pair can be added
phonebook['David'] = '111-222-3333'
print(phonebook)
````

````python
# Key-value pair can be removed using the del statement
del phonebook['Charlie']
print(phonebook)
````

````python
# Two dictionaries can be merged using the update() method
more_contacts = {
    'Eve': '444-444-4444',
    'Frank': '333-333-3333',
}
phonebook.update(more_contacts)
print(phonebook)

# # Alternatively, the merge operator can be used in Python 3.9+
# phonebook |= more_contacts
````

````python
# Any immutable type can be used as a key, including tuples
phonebook[('Grace', 'Hopper')] = '222-222-2222'
print(phonebook[('Grace', 'Hopper')])  # Output: 222-222-2222
````

# Množina

Množina je neuspořádaná a měnitelná kolekce, která neumožňuje duplicitní prvky. Definuje se uzavřením posloupnosti položek oddělených čárkami do složených závorek `{}` nebo pomocí funkce `set()`.

```python
my_set = {1, 2, 3, 'apple', 'banana', 3.14, True}
# or
my_set = set([1, 2, 3, 'apple', 'banana', 3.14, True])
```

````python
vegetables = {'carrot', 'broccoli', 'tomato'}
````

````python
# Adding an element to a set
vegetables.add('potato')
print(vegetables)
````

````python
# Removing an element from a set
vegetables.remove('broccoli')
print(vegetables)
````

````python
# Adding an existing element has no effect, as sets do not allow duplicates
vegetables.add('potato')
print(vegetables)
````

# Důležité funkce

Při práci s datovými strukturami se mohou hodit následující funkce.
Pro ukázky si definujme příklad seznamu a slovníku.

````python
# Define a list.
my_list = [10, 20, 30, 40, 50]

# Define a dictionary.
my_dict = {
    'name': 'Alice',
    'age': 30,
    'city': 'Wonderland'
}
````

`len()`: Vrací počet položek v kolekci.

````python
# Number of elements in the list
list_length = len(my_list)
print(f'Length of the list: {list_length}')  # Output: 5

# Number of elements (key-value pairs) in the dictionary
dict_length = len(my_dict)
print(f'Length of the dictionary: {dict_length}')  # Output: 3
````

`type()`: Vrací typ kolekce.

````python
# Type of the list
print('Type of my_list:', type(my_list))  # Output: <class 'list'>

# Type of the dictionary
print('Type of my_dict:', type(my_dict))  # Output: <class 'dict'>
````

`print()`: Vypíše obsah proměnné, včetně kolekce, do konzole.

````python
# Print the contents of the list
print('Contents of my_list:', my_list)  # Output: [10, 20, 30, 40, 50]

# Print the contents of the dictionary
print('Contents of my_dict:', my_dict)  # Output: {'name': 'Alice', 'age': 30, 'city': 'Wonderland'}
````

`range()`: Vytváří posloupnost čísel, která se často používá ve smyčkách.

````python
# `range` parameters are: start, stop, step
#  - start: the first number in the sequence (inclusive)
#  - stop: the last number in the sequence (exclusive)
#  - step: the difference between consecutive numbers in the sequence

# The output of the following `for` loop will be: 1 3 5 7 9
start = 1
stop = 11
step = 2
for i in range(start, stop, step):
    print(i, end = ' ')
````

Pomocí klíčového slova `in` ověříte, zda se položka nachází v n-tici, slovníku (kontrolují se klíče) nebo množině.

```python

```

# Iterace přes n-tici, slovník a množinu

Pomocí cyklu `for` můžete procházet položky n-tice, slovníku (procházejí se klíče) nebo množiny.

````python
# For dictionaries, the `for` loop iterates over keys
for name in phonebook:
    print(name, phonebook[name], end = ' ')
````

````python
# For dictionaries, the `items()` method can be used to get both keys and values
for name, number in phonebook.items():
    print(name, number, end = ' ')
````

# Časová složitost běžných operací se slovníky a množinami

Slovníky a množiny
- Přístup: O(1)
- Vyhledávání: O(1)
- Vkládání: O(1)
- Mazání: O(1)

Slovníky a množiny jsou implementovány pomocí hašovacích tabulek, které pro přístup, vyhledávání, vkládání a mazání poskytují průměrnou časovou složitost O(1).
Proto jsou slovníky a množiny při operacích zahrnujících vyhledávání, vkládání nebo mazání prvků obecně efektivnější než seznamy.


# Výrazy pro tvorbu kolekcí

Stručný způsob, jak vytvářet slovníky a množiny.

````python
# Dictionary comprehension
squared_dict = {x: x ** 2 for x in range(1, 6)}
print(squared_dict)  # Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Another example of dictionary comprehension
square_cube_dict = {x ** 2: x ** 3 for x in squared_dict}
print(square_cube_dict)
````

````python
# Set comprehension also exists
squares_set = {x**2 for x in range(1, 6)}
print(squares_set)
````
