# Podmínky a cykly


### Podmínka if

Příkaz `if` slouží ke spuštění bloku kódu pouze tehdy, když je zadaná podmínka pravdivá.
Blok kódu uvnitř příkazu `if` je odsazen vůči slovu `if`.
Odsazení se nejčastěji vytváří vložením čtyř mezer na začátek řádku.
Některá integrovaná vývojová prostředí (IDE) lze nastavit tak, aby po stisknutí klávesy Tab automaticky vložila požadované čtyři mezery.
Ověřte si, že to tak ve vašem IDE funguje, protože používání znaku Tab místo mezer se příliš nedoporučuje (i když funguje).

Za příkazem `if` mohou následovat příkazy `elif` (žádný nebo několik), které slouží ke kontrole více podmínek.
Pokud je podmínka v příkazu `if` nepravdivá, program postupně kontroluje podmínky v příkazech `elif`.
Posledním příkazem může být volitelný příkaz `else`, který se provede, pokud není pravdivá žádná z předchozích podmínek.

Předpokládejme například, že zjišťujeme, zda je číslo kladné, záporné, nebo nulové.
V příkazu `if` můžeme kontrolovat například podmínku `a > 0`, v příkazu `elif` podmínku `a < 0` a příkaz `else` použít pro případ, kdy je `a` rovno nule.
Kód by vypadal takto:

```python
if x > 0:
    print("Number x is positive.")
elif x < 0:
    print("Number x is negative.")
else:
    print("Number x is zero.")
```

Počet řádků ani složitost bloku kódu uvnitř příkazů `if`, `elif` a `else` nejsou omezeny.
Uvnitř těchto bloků mohou být také další vnořené příkazy `if`.

````python
# minimal example

a = 1
if a > 0:
    print("a is positive")
````

````python
# full example if-elif-else
# also demonstrates nested if statements

if (a == 1):
    print("hit!")
    a = 1000
    print(a)
    if (a == 1000):
        print('double hit!')
    print('ahoj')
elif (a == 0):
    print("bad luck")
elif (a == 2345):
    a = 0
    print(a)
else:
    print("try another time")

````

### Cyklus while

Cyklus `while` opakuje předem určený blok kódu, dokud je zadaná podmínka pravdivá.
Blok kódu uvnitř cyklu `while` je odsazen vůči slovu `while`.
Podmínka se uvádí hned za slovem `while`.
Nejprve se zkontroluje ve chvíli, kdy vykonávání kódu dospěje k příkazu `while`.
Pokud je pravdivá, blok kódu se vykoná.
Poté se podmínka zkontroluje znovu.
Tento postup pokračuje, dokud se podmínka nestane nepravdivou.
Jakmile k tomu dojde, program pokračuje prvním řádkem kódu za blokem `while` (tedy řádkem, který není odsazen vůči příkazu `while`).


````python
# The following code prints numbers from 1 to 9 using a while cycle.
# In each iteration, the value of `i`` is increased by 1.
# When `i` reaches 10, the condition `i < 10` becomes false, and the cycle ends.
# After that, the program continues with the first line after the cycle, which prints "done".

i = 1
while i < 10:
    print(i)
    i += 1
print("done")
````

### Cyklus for

Cyklus `for` se obvykle používá, když předem víme, kolikrát chceme blok kódu opakovat.
V nejjednodušším případě můžeme cyklus `for` použít k procházení posloupnosti čísel vytvořené funkcí `range()`.
Cyklus `for` začíná slovem `for`, po kterém následuje název proměnné, slovo `in` a posloupnost, kterou chceme procházet.
Blok kódu uvnitř cyklu `for` je odsazen vůči slovu `for`.
Při každém průchodu cyklem nabývá proměnná (`i` v příkladu níže) další hodnoty z posloupnosti a vykoná se blok kódu.
Po zpracování všech hodnot z posloupnosti program pokračuje prvním řádkem kódu za blokem `for` (tedy řádkem, který není odsazen vůči příkazu `for`).

````python
# Iterate over numbers from 0 to 9 and print only the even ones.
# Note that the `range(n)` function generates numbers from `0` to `n - 1`.
# The line `print("done")` is not indented with respect to the `for` statement,
# so it is executed only after the cycle ends.

for i in range(10):
    if i % 2 == 0:
        print(i)
print("done")
````

# Časová složitost programů

Časovou složitost nejlépe odhadneme podle počtu základních operací, které musí program provést.
Základní operací může být například součet dvou čísel, jejich porovnání nebo přístup k prvku seznamu pomocí jeho indexu.
Základní operací může být i posloupnost takových úkonů.
Časová složitost se obvykle vyjadřuje pomocí značení Big O, které popisuje horní mez složitosti v závislosti na velikosti vstupních dat (často označované jako `n`).
- O(1) = konstantní čas (doba jedné základní operace)
- O(n) = lineární čas (doba až n základních operací)

Pokud cyklus `for` prochází `n` položek a uvnitř cyklu se provádí pouze konstantní počet základních operací, je časová složitost O(n).
To znamená, že se určitá operace provede n-krát a její doba nezávisí na n.
Pokud je však uvnitř prvního cyklu další cyklus `for`, který také prochází `n` položek, je časová složitost O($n^2$), protože se nyní určitá operace s konstantní dobou provede $n^2$-krát.
Je-li číslo `n` malé, nemusíme si s časovou složitostí dělat starosti.
Pokud je však `n` velké (například v řádu milionů), je mimořádně důležité vyhnout se algoritmům se složitostí O($n^2$).

# Shrnutí

Podmínky a cykly jsou základními programovacími konstrukcemi, které umožňují rozhodování a opakované vykonávání bloků kódu.

- Příkaz `if` umožňuje podmíněné vykonávání podle toho, zda je zadaná podmínka pravdivá, nebo nepravdivá.
- Cyklus `while` umožňuje opakovaně vykonávat blok kódu, dokud zadaná podmínka zůstává pravdivá.
- Cyklus `for` slouží k procházení posloupnosti hodnot a vykoná blok kódu pro každou hodnotu v posloupnosti.

Pozor na vnořování cyklů, které může vést k vysoké časové složitosti (například O($n^2$)) a výrazně zhoršit výkon při velkých vstupech.
