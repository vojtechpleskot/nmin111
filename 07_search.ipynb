# Searching algorithms

- Linear Search: O(n)
    - Check each element one by one until the target is found.
    - Useful for unsorted data.
    - Invoked when using the `in` keyword on lists or tuples in Python.
- Binary Search: O(log n) (requires sorted data)
    - Divide the search interval in half repeatedly.

```python
# Implement the linear search algorithm yourselves in a function called linear_search(arr, target).
# The function returns `None` if the target is not found, or the index of the target if it is found.
```

```python
# Binary Search Implementation
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        guess = arr[mid]

        if guess == target:
            return mid
        if guess > target:
            high = mid - 1
        else:
            low = mid + 1

    return None

# Example usage:
N = 100000000
sorted_list = [i for i in range(N)]
print(binary_search(sorted_list, 3))  # Output: 3
print(binary_search(sorted_list, -1)) # Output: None
```
