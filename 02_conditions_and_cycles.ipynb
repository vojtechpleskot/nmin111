# Conditions and cycles


### if condition

The `if` statement is used to execute a block of code only if a specified condition is true.
The block of code inside the `if` statement is indented with respect to the word `if`.
Indentation is most commonly achieved by inserting four spaces at the beginning of the line.
Some Integrated Development Environments (IDEs) can be setup to automatically insert the required four spaces when you press the Tab key.
Make sure, it's the case for your IDE, as using the Tab character instead of spaces is not highly recommended (although it works).

The `if` statement can be followed by `elif` statements (zero or several) to check multiple conditions.
If the condition in the `if` statement is false, the program checks the conditions in the `elif` statements one by one.
The final statement can be an `else` statement (optional) that will be executed if none of the previous conditions are true.

As an example, let's assume we are checking whether a number is positive, negative, or zero.
We can check e.g. the condition `a > 0` in the `if` statement, the condition `a < 0` in the `elif` statement, and use the `else` statement to cover the case when `a` is zero.
The code would look like this:

```python
if x > 0:
    print("Number x is positive.")
elif x < 0:
    print("Number x is negative.")
else:
    print("Number x is zero.")
```

The code block inside the `if`, `elif`, and `else` statements has no limit on the number of lines or their complexity.
There can be additional `if` statements nested inside the blocks as well.

````python
# minimal example

a = 1
if a > 0:
    print("a is positive")
````

````python
# full example if-elif-else
# also demonstrates nested if statements

if (a == 1):
    print("hit!")
    a = 1000
    print(a)
    if (a == 1000):
        print('double hit!')
    print('ahoj')
elif (a == 0):
    print("bad luck")
elif (a == 2345):
    a = 0
    print(a)
else:
    print("try another time")

````

### while cycle

The `while` cycle repeats a predefined block of code as long as a specified condition is true.
The block of code inside the `while` cycle is indented with respect to the word `while`.
The condition is specified right after the word `while`.
It is checked when the code execution reaches the `while` statement, first.
If it's true, the block of code is executed.
After that, the condition is checked again.
This process continues until the condition becomes false.
When that happens, the program execution continues with the first line of code after the `while` block (i.e., the line that is not indented with respect to the `while` statement).


````python
# The following code prints numbers from 1 to 9 using a while cycle.
# In each iteration, the value of `i`` is increased by 1.
# When `i` reaches 10, the condition `i < 10` becomes false, and the cycle ends.
# After that, the program continues with the first line after the cycle, which prints "done".

i = 1
while i < 10:
    print(i)
    i += 1
print("done")
````

### for cycle

The `for` cycle is typically used when we know in advance how many times we want to repeat a block of code.
In the simplest case, we can use the `for` cycle to iterate over a sequence of numbers generated by the `range()` function.
The `for` cycle starts with the word `for`, followed by a variable name, the word `in`, and the sequence to iterate over.
The block of code inside the `for` cycle is indented with respect to the word `for`.
In each iteration of the cycle, the variable (`i` in the example below) takes the next value from the sequence, and the block of code is executed.
After all values in the sequence have been used, the program execution continues with the first line of code after the `for` block (i.e., the line that is not indented with respect to the `for` statement).

````python
# Iterate over numbers from 0 to 9 and print only the even ones.
# Note that the `range(n)` function generates numbers from `0` to `n - 1`.
# The line `print("done")` is not indented with respect to the `for` statement,
# so it is executed only after the cycle ends.

for i in range(10):
    if i % 2 == 0:
        print(i)
print("done")
````

# Time complexity of programs

Time complexity is best estimated with the number of some elementary operations that your program has to perform.
An elementary operation can be, for example, an addition of two numbers, a comparison of two numbers, or accessing an element in a list by its index.
It can also be a sequence of such operations.
Time complexity is usually expressed using Big O notation, which describes the upper bound of the time complexity in terms of the size of the input data (often denoted as `n`).
- O(1) = constant time (time of one elementary operation)
- O(n) = linear time (time of up to n elementary operations)

If you have a `for` cycle that goes through `n` items, and inside the cycle you have only a constant number of elementary operations, then the time complexity is O(n).
It means that some operation is performed n times, and the time of the operation does not depend on n.
However, if there is another `for` cycle inside the first one, also going through `n` items, then the time complexity is O($n^2$), because now some constant-time operation is performed $n^2$ times.
When the number `n` is small, one doesn't need to worry about time complexity.
However, if `n` is large (e.g., millions), then it is crucially important to avoid O($n^2$) algorithms.

# Summary

Conditions and cycles are fundamental constructs in programming that allow for decision-making and repetitive execution of code blocks.

- The `if` statement enables conditional execution based on whether a specified condition is true or false.
- The `while` cycle allows for repeated execution of a code block as long as a specified condition remains true.
- The `for` cycle is used for iterating over a sequence of values, executing a code block for each value in the sequence.

Be careful with nesting cycles, as they can lead to high time complexity (e.g., O($n^2$)), which can significantly impact performance for large input sizes.
