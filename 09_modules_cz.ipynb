# Moduly

- Moduly se v Pythonu používají k rozdělení kódu do přehledných částí.

- Modul je soubor obsahující definice a příkazy jazyka Python.

- Název modulu je název souboru doplněný příponou `.py`.

- Modul se importuje příkazem `import`.

- Některé moduly jsou vestavěné, jiné mohou vytvářet uživatelé nebo je lze instalovat z externích zdrojů.

- Moduly lze pro lepší organizaci seskupovat do balíčků.


````python
# Import the built-in math module
import math

# Use a function from the math module
result = math.sqrt(16)
print(result)
````

````python
# Import the math module with an alias
import math as m
result = m.sqrt(25)
print(result)
````

````python
# Import the function sqrt directly from the math module
from math import sqrt

# When calling the function, no module prefix is needed
result = sqrt(36)
print(result)
````

# Balíčky

- Balíček slouží k organizaci souvisejících modulů.

- Balíček je adresář, který obsahuje speciální soubor s názvem `__init__.py` a další soubory modulů.

- Příkazem `import A.B` importujete modul `B` z balíčku `A`.

- Balíčky často mají hierarchickou strukturu s podbalíčky.

  - Pomocí `import A.B.C` importujete modul `C` z podbalíčku `B` balíčku `A`.

```python
# Example of importing a module from a package
import matplotlib.pyplot
matplotlib.pyplot.plot([1, 2, 3], [4, 5, 6])
matplotlib.pyplot.show()
```


# Spouštění modulů jako skriptů

- Moduly lze spouštět jako samostatné skripty z příkazového řádku.

- Konstrukci `if __name__ == "__main__":` použijte pro kód, který se má spustit pouze tehdy, když je modul spuštěn jako skript.

  - Poznámka: uvnitř modulu je proměnná `__name__` nastavena na název modulu. Pokud je však modul spuštěn jako hlavní program, `__name__` je nastavena na `"__main__"`.

- To se hodí při testování nebo předvádění funkcí modulu.

- Příklad:
  ```python
  def main():
      print("This module is being run directly")

  if __name__ == "__main__":
      main()
  ```


# Vyhledávací cesta modulů

- Při importu modulu s názvem `xyz` hledá interpret nejprve vestavěný modul s tímto názvem.

- Pokud jej nenajde, prohledává seznam adresářů určený proměnnou sys.path a hledá v něm soubor s názvem `xyz.py`. Proměnná sys.path se inicializuje z těchto umístění:

  - Adresář obsahující vstupní skript (nebo aktuální adresář, pokud nebyl zadán žádný soubor).

  - PYTHONPATH (seznam názvů adresářů se stejnou syntaxí jako proměnná prostředí PATH).

  - Výchozí umístění závislá na instalaci (obvykle zahrnuje adresář site-packages, jehož zpracování zajišťuje modul site).


# Stahování modulů a balíčků

- Z Python Package Indexu (PyPI) lze stáhnout mnoho modulů a balíčků třetích stran.

- K jejich snadné instalaci použijte nástroje pro správu balíčků, například `pip`.

  - Příklad příkazu pro instalaci balíčku:
    ```
    pip install package_name
    ```

- Dalším oblíbeným nástrojem je `conda`, který se často používá v prostředích pro datovou vědu a vědecké výpočty.

  - Příklad instalace balíčku pomocí conda:
    ```
    conda install package_name
    ```


# Některé vestavěné moduly


````python
# Random number generation
import random
random_number = random.randint(1, 10)
print("Random number between 1 and 10:", random_number)

# Uniform distribution
uniform_number = random.uniform(0, 1)
print("Random float between 0 and 1:", uniform_number)

# Gaussian distribution
gaussian_number = random.gauss(0, 1)
print("Random number from Gaussian distribution (mean=0, stddev=1):", gaussian_number)
````

````python
# timer
import time
start_time = time.time()
# Some code to time
time.sleep(1)  # Sleep for 1 second
end_time = time.time()
elapsed_time = end_time - start_time
print("Elapsed time:", elapsed_time, "seconds")
````

````python
# Date and time manipulation
import datetime
current_datetime = datetime.datetime.now()
print("Current date and time:", current_datetime)

# Just the date
current_date = datetime.date.today()
print("Current date:", current_date)
````

# Jmenné prostory

- Jmenný prostor je kontejner obsahující množinu identifikátorů (názvů) a jim odpovídajících objektů.

- Různé moduly mají vlastní jmenné prostory, což pomáhá předcházet kolizím názvů.

- K objektům ve jmenném prostoru modulu lze přistupovat pomocí tečkové notace (např. `module_name.object_name`).

- Příklady jmenných prostorů:
  
  - globální názvy v modulu
  
  - lokální názvy při volání funkce
  
  - množina vestavěných názvů (obsahuje funkce, jako je print())

- Rozsah platnosti označuje viditelnost názvů v programu.

  - Například název definovaný uvnitř funkce není mimo tuto funkci viditelný.


````python
# The same name can be used when it is in different namespaces
def sqrt(x):
    return x ** 0.5

import math

print(sqrt(9), math.sqrt(9))
````
