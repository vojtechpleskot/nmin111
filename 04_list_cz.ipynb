# Seznam

- Seznam je uspořádaná kolekce hodnot.
- Jednotlivé prvky mohou být libovolného typu.
- Prvky jsou indexovány celými čísly počínaje 0.
- Seznamy mohou obsahovat duplicitní prvky.
- Seznam je měnitelný, což znamená, že jej lze po vytvoření upravovat.
    - Prvky lze přidávat, odebírat nebo upravovat.
- Seznam se zapisuje tak, že se posloupnost položek oddělených čárkami uzavře do hranatých závorek `[]`.

```python
my_list = [1, 2, 3, 'apple', 'banana', 3.14, True]
```

````python
# Create a list of fruits
fruits = ['apple', 'banana', 'cherry', 'banana']
````

````python
# Elements are accessed by their index (starting from 0), using square brackets
print(fruits[0])  # Output: apple
print(fruits[1])  # Output: banana
````

````python
# Indexing can also be negative, where -1 refers to the last item
print(fruits[-1])  # Output: banana
print(fruits[-2])  # Output: cherry
````

````python
# Adding an element to the end of the list
fruits.append('orange')
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'banana', 'orange']
````

````python
# Removing the last element from the list
fruits.pop()
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'banana']
````

````python
# Elements can be updated by assigning a new value to an existing index
fruits[1] = 'blueberry'
print(fruits)  # Output: ['apple', 'blueberry', 'cherry', 'banana']
````

````python
# Adding two lists
more_fruits = ['kiwi', 'mango']
all_fruits = fruits + more_fruits
print(all_fruits)  # Output: ['apple', 'banana', 'cherry', 'banana', 'kiwi', 'mango']
````

````python
# Operator +=
fruits += ['grape']
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'banana', 'grape']
````

# Procházení seznamu

K procházení položek seznamu použijte cyklus `for`.

```python
for fruit in fruits:
    print(fruit)
```

````python
# Iterate over the list `fruits`
for fruit in fruits:
    print(fruit, end = ' ')
````

````python
# For lists and tuples, the `enumerate()` function can be used to get both index and value
for index, fruit in enumerate(fruits):
    print(index, fruit)
````

````python
# One can iterate over several lists in parallel using the `zip()` function
cost = [50, 75, 60, 40, 30]
for fruit, price in zip(fruits, cost):
    print(fruit, price, end = ' ')
````

# Seznamová komprehenze

Stručný způsob, jak vytvořit seznam.

````python
# List comprehension
integers = [x for x in range(1, 6)]

# Another example of list comprehension
squares = [x**2 for x in integers]
print(squares)  # Output: [1, 4, 9, 16, 25]
````

Seznamové komprehenze mohou obsahovat také podmínky. Zápis používá klíčové slovo `if`.

````python
# Filtering with list comprehension
even_squares = [x**2 for x in integers if x % 2 == 0]
````

# Časová složitost programů

Časovou složitost nejlépe odhadneme podle počtu základních operací, které musí program provést.
Základní operací může být například součet dvou čísel, jejich porovnání nebo přístup k prvku seznamu pomocí jeho indexu.
Základní operací může být i posloupnost takových úkonů.
Časová složitost se obvykle vyjadřuje pomocí značení Big O, které popisuje horní mez složitosti v závislosti na velikosti vstupních dat (často označované jako `n`).
- O(1) = konstantní čas (doba jedné základní operace)
- O(n) = lineární čas (doba až n základních operací)

Pokud cyklus `for` prochází `n` položek a uvnitř cyklu se provádí pouze konstantní počet základních operací, je časová složitost O(n).
To znamená, že se určitá operace provede n-krát a její doba nezávisí na n.
Pokud je však uvnitř prvního cyklu další cyklus `for`, který také prochází `n` položek, je časová složitost O($n^2$), protože se nyní určitá operace s konstantní dobou provede $n^2$-krát.
Je-li číslo `n` malé, nemusíme si s časovou složitostí dělat starosti.
Pokud je však `n` velké (například v řádu milionů), je mimořádně důležité vyhnout se algoritmům se složitostí O($n^2$).

# Časová složitost běžných operací se seznamem

- Přístup: O(1)
- Vyhledávání: O(n)
- Přidání na konec: O(1)
- Odebrání: O(1) (při odebrání z konce)
- Vložení na libovolnou pozici: O(n)
- Smazání z libovolné pozice: O(n)

Vložení nebo smazání prvku na začátku či uprostřed seznamu má složitost O(n), protože je nutné posunout všechny následující prvky.
Seznamy nejsou optimalizovány pro rychlé vkládání a mazání na libovolných pozicích.
Pokud záleží na výkonu (například u velkých seznamů), je vhodné se těmto operacím vyhnout.
Vyhledávání prvku v seznamu má také složitost O(n), protože v nejhorším případě může být nutné zkontrolovat každý prvek.

### Eratosthenovo síto

Algoritmus je popsán zde: https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes.
Převzato ze stránky Wikipedie:

 - Vytvořte seznam po sobě jdoucích celých čísel od 2 do n: (2, 3, 4, ..., n).

 - Na začátku nastavte p na 2, nejmenší prvočíslo.

 - Vypište násobky p tak, že budete přičítat p od 2p až do n, a označte je v seznamu (půjde o 2p, 3p, 4p, ...; samotné p neoznačujte).

 - Najděte nejmenší neoznačené číslo v seznamu, které je větší než p. Pokud takové číslo neexistuje, skončete. Jinak nastavte p na toto nové číslo (další prvočíslo) a opakujte od kroku 3.
 
 - Po skončení algoritmu jsou všechna neoznačená čísla v seznamu prvočísla menší než n.

Poznámky:

 - Můžete použít druhý seznam, který bude pro každé číslo obsahovat logickou hodnotu a umožní sledovat, která čísla jsou označena.

 - Můžete využít funkci `range` a její parametr `step`.

```python
start = 0
stop  = 10
step  = 2
for i in range(start, stop, step):
  print(i, end = ' ')
# output will be: 0 2 4 6 8
```

 - Eratosthenovo síto je efektivní algoritmus pro vypracování domácího úkolu. Nezapomeňte, že:

   - Vaše funkce se musí jmenovat `prvocisla(n)`.

   - Váš soubor se musí jmenovat `reseni.py`.
   
   - Ve zdrojovém kódu nevolejte žádnou funkci. Zejména nevolejte funkci `input()`.
