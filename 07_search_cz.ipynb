# Vyhledávací algoritmy

- Lineární vyhledávání: O(n)
    - Postupně kontroluje každý prvek, dokud nenajde hledanou hodnotu.
    - Hodí se pro neuspořádaná data.
    - Použije se při použití klíčového slova `in` se seznamy nebo n-ticemi v Pythonu.
- Binární vyhledávání: O(log n) (vyžaduje seřazená data)
    - Opakovaně rozdělí interval vyhledávání na poloviny.

```python
# Binary Search Implementation
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        guess = arr[mid]

        if guess == target:
            return mid
        if guess > target:
            high = mid - 1
        else:
            low = mid + 1

    return None

# Example usage:
N = 100000000
sorted_list = [i for i in range(N)]
print(binary_search(sorted_list, 3))  # Output: 3
print(binary_search(sorted_list, -1)) # Output: None
```
