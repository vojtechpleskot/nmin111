# Instalace Pythonu na Windows 11

- Krok za krokem návod připravený 12.2.2026. Pro začátek jděte na [python.org/downloads/windows](https://www.python.org/downloads/windows/).

- V případě potřeby se neváhejte ptát nástrojů jako ChatGPT nebo Gemini.

- Stáhněte si verzi Pythonu dle vlastního výběru. Pokud si nejste jisti, stáhněte si nejnovější stabilní verzi.



- Po stažení spusťte instalační program. V Google Chrome stačí kliknout na stažený soubor v pravém horním rohu.



- Zaškrtněte políčko "Add python.exe to PATH" ve spodní části první instalační obrazovky. To vám umožní spouštět Python z příkazového řádku.



- Poté klikněte na "Install Now".



- Po úspěšné instalaci uvidíte následující okno. Klikněte na "Close".



# Instalace Visual Studio Code na Windows 11

- Krok za krokem návod připravený 12.2.2026. Stačí následovat níže uvedené snímky obrazovky a věnovat pozornost červeným rámečkům. Pro začátek jděte na [code.visualstudio.com/download](https://code.visualstudio.com/download). Klikněte na tlačítko "Windows" pro stažení instalačního programu.



- Po stažení spusťte instalační program. V Google Chrome stačí kliknout na stažený soubor v pravém horním rohu.



- Přijměte licenční smlouvu a klikněte na "Next".



- Zvolte instalační složku a klikněte na "Next". Ponechte výchozí umístění, pokud nemáte specifický důvod ke změně.



- Vyberte složku v nabídce Start a klikněte na "Next". Ponechte výchozí název, pokud nemáte specifický důvod ke změně.



- Zaškrtněte tři políčka, jak je znázorněno na snímku obrazovky níže. Klikněte na "Next".



- Klikněte na "Install" pro spuštění instalace.



- Po úspěšné instalaci uvidíte následující okno. Klikněte na "Finish".

# Nastavení VS Code, včetně instalace GitHub Copilot

- Když otevřete VS Code poprvé, uvidíte uvítací obrazovku. Klikněte na "Use AI Features" v sekci "Get Started with VS Code". Následující průvodce krok za krokem začne od tohoto bodu. Pokud jste již uvítací obrazovku zavřeli, nevadí. Ke stejným možnostem můžete přistoupit kliknutím na ikonu "Extensions" v levém postranním panelu (vypadá jako čtyři čtverce), vyhledáním "GitHub Copilot Chat" a kliknutím na "Install" pro toto rozšíření (viz snímek obrazovky v [Instalace GitHub Copilot Chat](#instalace-github-copilot-chat) níže).



- Budete požádáni o přihlášení ke svému účtu GitHub. Klikněte na "Sign in with GitHub" a postupujte podle pokynů k ověření vašeho účtu. Ujistěte se, že používáte stejný účet GitHub, pro který máte bezplatnou licenci Copilot.



- K vašemu účtu GitHub se můžete přihlásit například pomocí svého Google účtu, jak je uvedeno na snímku obrazovky níže.



- Když je ověření úspěšné, uvidíte obrazovku jako tu níže. Vše je nastaveno a můžete začít používat funkce uvedené na snímku obrazovky!



- Nainstalujte rozšíření "Python" pro VS Code. Klikněte na ikonu "Extensions" v levém postranním panelu (vypadá jako čtyři čtverce), vyhledejte "Python" a klikněte na "Install" pro toto rozšíření.



- Další rozšíření, které budete potřebovat, je "Jupyter". Toto rozšíření poskytuje podporu pro Jupyter Notebooky ve VS Code. Jupyter notebooku budeme v tomto kurzu používat velmi intenzivně. Vyhledejte "Jupyter" v Extensions a klikněte na "Install".



### Instalace GitHub Copilot Chat

- Klikněte na ikonu "Extensions" v levém postranním panelu (vypadá jako čtyři čtverce), vyhledejte "GitHub Copilot Chat" a klikněte na "Install" pro toto rozšíření.



- Poté postupujte podle pokynů k přihlášení ke svému účtu GitHub a ověření rozšíření. Ujistěte se, že používáte stejný účet GitHub, pro který máte bezplatnou licenci GitHub Copilot.

# První projekt v Pythonu ve VS Code

- Když otevřete VS Code, můžete otevřít nebo vytvořit nový projekt. Projekt je jednoduše složka na vašem počítači, kde budete ukládat své soubory se zdrojovým kódem. Můžete vytvořit novou složku pro váš projekt, nebo můžete otevřít existující složku, pokud již nějakou máte. Klikněte na "Open folder" (Otevřít složku) a vyberte složku, kterou chcete použít pro váš projekt.



- Pokud ještě nemáte složku, můžete vytvořit novou kliknutím na "New folder" (Nová složka) v průzkumníku souborů.



- Potvrďte, že důvěřujete autorům souborů v projektu.



- Vytvořte nový pythonní soubor ve vašem projektu kliknutím na ikonu "New File" (Nový soubor) v průzkumníku souborů a pojmenujte jej s příponou .py, například "module1.py".



- Začněte popisem účelu modulu v docstringu. Pokud je GitHub Copilot Chat správně nainstalován a nastaven, měl by začít navrhovat text pro docstring hned, jak začnete psát. Návrhy můžete přijmout stisknutím klávesy "Tab".



- Běžný způsob používání GitHub Copilot je napsat docstring nebo komentář a nechat Copilot navrhnout kód za vás. Následující kód byl vygenerován GitHub Copilotem:



- Pokud má váš modul spustitelnou část, můžete ji otestovat jednoduchým kliknutím na ikonu "Run" (Spustit - trojúhelník) v pravém horním rohu editoru. Tím se spustí aktuální soubor a zobrazí se výstup v terminálu. Musíte určit interpret Pythonu pro spuštění kódu. Pokud nemáte nastavené virtuální prostředí, můžete vybrat globální interpret Pythonu, který jste nainstalovali dříve.



- Pro otevření terminálu ve VS Code jednoduše přetáhněte dolní panel nahoru a pak klikněte na záložku "Terminal". V terminálu můžete spouštět příkazy a skripty Pythonu, jak je ukázáno na obrázku níže.



# Klávesové zkratky

- Seznam užitečných klávesových zkratek pro VS Code najdete [zde](https://code.visualstudio.com/shortcuts/keyboard-shortcuts-windows.pdf). Naučení a používání klávesových zkratek může výrazně zlepšit vaši produktivitu a zpříjemnit zkušenost se psaním zdrojového kódu.

| Klávesová zkratka | Akce |
|-|-|
| `Ctrl + C` | Kopírovat vybraný text nebo kód. |
| `Ctrl + V` | Vložit zkopírovaný text nebo kód. |
| `Ctrl + X` | Vyjmout vybraný text nebo kód. |
| `Ctrl + F` | Najít v aktuálním souboru. |
| `Ctrl + Shift + F` | Najít v souborech |
| `Ctrl + Z` | Vrátit poslední akci. |
| `Ctrl + Y` | Zopakovat poslední vrácenou akci. |
| `Ctrl + Alt + Up/Down` | Víceřádkový kurzor. Toto "protáhne" váš kurzor svisle nahoru nebo dolů.
| `Ctrl + D` | Vícenásobný výběr. Označí další výskyt vybraného slova, takže můžete najednou upravit více věcí. |
| `Ctrl + /` | Zakomentovat nebo odkomentovat vybrané řádky. |
| `Ctrl + ]/[` | Odsadit nebo zrušit odsazení vybraných řádků. |
| `Alt + Up/Down` | Přesunout řádek nahoru/dolů |
| F12 | Přejít na definici |
| `Ctrl + G` | Přejít na řádek. Zadejte číslo pro skok přímo na daný řádek. |
