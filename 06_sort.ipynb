# Sorting

Goal: Sort numbers (or some other elements) in a predefined order (e.g. in ascending order).

The goal is simple but the solution is not, as the time complexity can be large.
The simplest sorting algorithms (e.g. bubble sort) have a time complexity of O(n^2), while the most advanced algorithms achieve O(n log n).

# Simplest Sorting Algorithms

Let's look at the simple O(n^2) sorting algorithms.

### Sorting by repeated selection of minimum


```python
# Sorting by repeated selection of minimum

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        # Assume the minimum is the first element of the unsorted part
        min_index = i
        # Iterate through the unsorted elements
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        # Swap the found minimum element with the first element of the unsorted part
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr

# Example usage
numbers = [64, 25, 12, 22, 11, -5]
sorted_numbers = selection_sort(numbers)
print("Sorted array:", sorted_numbers)
```

```python
# Let's experiment with high-n inputs
N = 10000
long_list = [i for i in range(N, 0, -1)]

# Set the timer.
import time
start_time = time.time()

# Sort the list using our selection sort implementation
sorted_long_list = selection_sort(long_list)

# Stop the timer.
end_time = time.time()

# Display the results.
print("Sorted large array:", sorted_long_list)
print("Time taken:", end_time - start_time)
```

### Bubble Sort

Bubble sort is a commonly-cited bad (i.e. O(n^2) again) sorting algorithm.
Description:
- Do the following `n`-times:
    - For each value of `i` from `0` to `n-2` do:
        - If the element at index `i` is greater than the element at index `i+1`, swap them.

# Real life sorting algorithms

- Never use Bubble Sort or any other O(n^2) sorting algorithm in real life!
- Also, don't reinvent the wheel: use built-in sorting functions of your programming language of choice!
- In Python, use the `sorted()` function or the `.sort()` method of lists.
    - They have a time complexity of O(n log n).

```python
# Sort the list using Python's built-in sorted function
start_time = time.time()
second_sorted_long_list = sorted(long_list)
end_time = time.time()
print("Sorted large array (built-in):", second_sorted_long_list)
print("Time taken (built-in):", end_time - start_time)
```

```python
# Sort in the descending order
short_list = [5, 3, 8, 1, 2]
descending_sorted_short_list = sorted(short_list, reverse = True)
print("Descending sorted short list:", descending_sorted_short_list)
```

# Sorting strings: lexicographical order

- Strings are sorted based on the Unicode code points of their characters.

- The Unicode code of "standard" characters is assigned in an intuitive way (e.g., 'a' < 'b' < 'c', 'A' < 'B' < 'C', '0' < '1' < '2').

- Uppercase letters have lower Unicode code points than lowercase letters (e.g., 'A' < 'a').

- Non-alphanumeric characters have varying Unicode code points, which can affect their sorting order.

- If the first characters of two strings are the same, the second characters are compared, and so on.

```python
# Sort list of names in the lexicographical order.
names = ['Charlie', 'Alice', 'Bob', 'David', 'Bill']
sorted_names = sorted(names)
print(sorted_names)
```

# Sorting with a custom key

- Use the `key` parameter of the `sorted()` function or the `.sort()` method to specify a custom sorting criterion.

- For example, you can sort a list of strings by their length.

- The `key` parameter takes a function that extracts a comparison key from each element in the collection.


```python
names = ['Charlie', 'Alice', 'Bob', 'David', 'Bill']
sorted_names = sorted(names, key = len)
print(sorted_names)  # Output: ['Bob', 'Bill', 'Alice', 'David', 'Charlie']
```

```python
# One can sort according to several criteria by using a tuple as the key.
# The lexicographical order is followed in the sense that the first tuple elements are compared first,
# and only if they are equal, the second elements are compared, etc.
people = [['Alice', 30], ['Bob', 25], ['Charlie', 30], ['David', 20]]

# Define the key function.
def sort_key(person):
    name = person[0]
    age = person[1]
    return (age, name)

# Sort by age, then by name
sorted_people = sorted(people, key = sort_key)
print(sorted_people) # Output: [('David', 20), ('Bob', 25), ('Alice', 30), ('Charlie', 30)]
```

```python
# Sort the people above from the oldest to the youngest; sort people with the same age according to their name alphabetically.
def sort_key_desc(person):
    name = person[0]
    age = person[1]
    return (-age, name)

sorted_people_desc = sorted(people, key = sort_key_desc)
print(sorted_people_desc)  # Output: [('Alice', 30), ('Charlie', 30), ('Bob', 25), ('David', 20)]
```

### Lambda functions as sorting keys

- You can use lambda functions as the `key` parameter to define custom sorting criteria inline.

```python
# Define the sort_key function using a lambda function.
sorted_people = sorted(people, key = lambda person: (person[1], person[0]))
print(sorted_people) # Output: [('David', 20), ('Bob', 25), ('Alice', 30), ('Charlie', 30)]
```
